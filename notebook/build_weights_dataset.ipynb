# Download Huihui-Qwen3.8-27B-Abliterated Weights on Kaggle Free CPU

This notebook runs on **free, unmetered Kaggle CPU** to download the BF16 safetensors weights (~55 GB) for `huihui-ai/Huihui-Qwen3.8-27B-abliterated` from Hugging Face.

### Instructions:
1. Accelerator: **CPU** (do NOT use TPU or GPU).
2. Internet: **On**.
3. Run all cells (~10-15 minutes).
4. In the right panel under **Output**, click `...` -> **Create Dataset**.
5. Set title to `huihui-qwen38-27b-abliterated` (slug: `huihui-qwen38-27b-abliterated`).

```python
import os
import time
from pathlib import Path
from huggingface_hub import snapshot_download

HF_REPO = "huihui-ai/Huihui-Qwen3.8-27B-abliterated"
OUTPUT_DIR = Path("/kaggle/working/weights")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Downloading {HF_REPO} to {OUTPUT_DIR}...")
t0 = time.time()
snapshot_download(
    repo_id=HF_REPO,
    local_dir=str(OUTPUT_DIR),
    allow_patterns=["*.safetensors", "*.json", "*.txt", "tokenizer*", "vocab*", "merges*"]
)
print(f"Done in {int(time.time() - t0)} seconds!")
print("Files in output:")
!ls -lh /kaggle/working/weights
```
